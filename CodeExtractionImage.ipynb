```python
import os
import requests
import cv2
import shutil

# Giả sử bạn đã có DataFrame df, headers và base_dir từ trước
# Lọc danh sách chỉ gồm những người có quốc tịch Nhật Bản
try:
    df_japan = df[df["nationality"] == "japan"]
except NameError:
    raise NameError("DataFrame 'df' chưa được định nghĩa. Vui lòng xác định 'df' chứa dữ liệu trước khi chạy script.")

# Thư mục lưu ảnh cho Japan (lưu vào japan2)
oj_dir = os.path.join(base_dir, "japan2")
os.makedirs(oj_dir, exist_ok=True)

# Load Haar Cascade
haar_xml_path = 'haarcascade_frontalface_default.xml'
face_cascade = cv2.CascadeClassifier(haar_xml_path)
if face_cascade.empty():
    raise FileNotFoundError(f"Không tìm thấy file Haar Cascade tại: {haar_xml_path}")

# Hàm crop vuông và resize về 160x160
def crop_and_resize_square(img, x, y, w, h, size=160):
    side = max(w, h)
    cx, cy = x + w // 2, y + h // 2
    x1 = max(cx - side // 2, 0)
    y1 = max(cy - side // 2, 0)
    x2 = min(x1 + side, img.shape[1])
    y2 = min(y1 + side, img.shape[0])
    x1 = max(x2 - side, 0)
    y1 = max(y2 - side, 0)
    square = img[y1:y2, x1:x2]
    resized = cv2.resize(square, (size, size), interpolation=cv2.INTER_AREA)
    return resized

# Nhóm các ảnh theo nhãn giống nhau trong df_japan
for idx, (label, group) in enumerate(df_japan.groupby("label"), start=1):
    save_dir = os.path.join(oj_dir, str(idx))
    os.makedirs(save_dir, exist_ok=True)
    valid_files = []

    for row in group.itertuples(index=False):
        image_url = row.image
        image_id = row.id
        try:
            response = requests.get(image_url, headers=headers, timeout=10)
            if response.status_code == 200:
                file_path = os.path.join(save_dir, f"{image_id}.jpg")
                with open(file_path, "wb") as file:
                    file.write(response.content)

                img = cv2.imread(file_path)
                if img is None:
                    print(f"Ảnh {file_path} bị hỏng, đã xóa.")
                    os.remove(file_path)
                    continue

                gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
                faces = face_cascade.detectMultiScale(
                    gray,
                    scaleFactor=1.1,
                    minNeighbors=5,
                    minSize=(30, 30),
                    flags=cv2.CASCADE_SCALE_IMAGE
                )

                if len(faces) == 0:
                    os.remove(file_path)
                    print(f"Xóa ảnh {file_path}: không có khuôn mặt")
                else:
                    x, y, w, h = faces[0]
                    face_resized = crop_and_resize_square(img, x, y, w, h, size=160)
                    cv2.imwrite(file_path, face_resized)

                    valid_files.append(file_path)
                    print(f"Giữ ảnh {file_path}: phát hiện {len(faces)} khuôn mặt")
            else:
                print(f"Lỗi tải ảnh {image_id}: {response.status_code}")
        except Exception as e:
            print(f"Lỗi khi xử lý ảnh {image_id}: {e}")

    if not valid_files:
        try:
            shutil.rmtree(save_dir, ignore_errors=True)
            print(f"Đã xóa thư mục {save_dir} (không có ảnh hợp lệ)")
        except Exception as e:
            print(f"Lỗi khi xóa thư mục {save_dir}: {e}")
    else:
        print(f"Thư mục {save_dir} giữ lại do có {len(valid_files)} ảnh hợp lệ")

print("Hoàn thành!")
```

```python
# Bước 1: Lấy danh sách các thư mục hiện có trong 'oj_dir'
existing_folders = [f for f in os.listdir(oj_dir) if os.path.isdir(os.path.join(oj_dir, f))]

# Bước 2: Sắp xếp thư mục theo số thứ tự (nếu có)
existing_folders.sort(key=lambda x: int(x) if x.isdigit() else float('inf'))

# Bước 3: Đánh số lại từ 1 và đổi tên thư mục
for new_idx, old_folder_name in enumerate(existing_folders, start=1):
    old_path = os.path.join(oj_dir, old_folder_name)
    new_path = os.path.join(oj_dir, str(new_idx))
    
    # Nếu tên thư mục đã khác với số mới, đổi tên
    if old_folder_name != str(new_idx):
        try:
            os.rename(old_path, new_path)
            print(f"Đã đổi tên {old_folder_name} → {new_idx}")
        except Exception as e:
            print(f"Lỗi khi đổi tên {old_folder_name}: {e}")

print("Đã đánh số lại thư mục xong!")
```

```python
import os
import cv2
import numpy as np

# Đường dẫn gốc chứa các thư mục label (hãy đảm bảo biến base_dir được định nghĩa trước đó)
root_dir = os.path.join(base_dir, "japan2")

def augment_image(img):
    """
    Trả về list các tuple (suffix, augmented_image)
    Mỗi ảnh gốc sẽ sinh 5 biến thể:
      - 1 flip ngang
      - 2 xoay ±15°
      - 1 điều chỉnh độ sáng
      - 1 thêm nhiễu Gaussian
    Tổng 5 ảnh.
    """
    h, w = img.shape[:2]
    aug_list = []

    # 1. Flip ngang
    aug_list.append(('_flip', cv2.flip(img, 1)))

    # 2. Xoay ±15°
    for angle in (-15, 15):
        M = cv2.getRotationMatrix2D((w // 2, h // 2), angle, 1.0)
        rot = cv2.warpAffine(img, M, (w, h), borderMode=cv2.BORDER_REFLECT)
        aug_list.append((f'_rot{angle}', rot))

    # 3. Thay đổi độ sáng (alpha > 1 để làm sáng nhẹ)
    bright = cv2.convertScaleAbs(img, alpha=1.2, beta=0)
    aug_list.append(('_bright', bright))

    # 4. Thêm nhiễu Gaussian
    gauss = np.random.normal(0, 15, img.shape).astype(np.float32)
    noisy = cv2.add(img.astype(np.float32), gauss)
    noisy = np.clip(noisy, 0, 255).astype(np.uint8)
    aug_list.append(('_noise', noisy))

    return aug_list


if __name__ == '__main__':
    for label_folder in os.listdir(root_dir):
        folder_path = os.path.join(root_dir, label_folder)
        if not os.path.isdir(folder_path):
            continue

        for fname in os.listdir(folder_path):
            if not fname.lower().endswith('.jpg'):
                continue

            img_path = os.path.join(folder_path, fname)
            img = cv2.imread(img_path)
            if img is None:
                print(f"Không đọc được ảnh {img_path}, bỏ qua.")
                continue

            for suffix, aug_img in augment_image(img):
                name, ext = os.path.splitext(fname)
                out_name = f"{name}{suffix}{ext}"
                out_path = os.path.join(folder_path, out_name)
                cv2.imwrite(out_path, aug_img)
                print(f"Tạo: {out_path}")

    print("Tăng cường dữ liệu hoàn tất!")
```

```python
import os
import cv2
import shutil
import stat

# Đường dẫn tuyệt đối đến thư mục japan2
root_dir = r"C:\Users\DELL\OneDrive\Documents\FaceDetection\Download_Images\japan2"

# Load Haar Cascade
cascade_path = 'haarcascade_frontalface_default.xml'
face_cascade = cv2.CascadeClassifier(cascade_path)
if face_cascade.empty():
    raise FileNotFoundError(f"Không tìm thấy file Haar Cascade tại: {cascade_path}")

def _handle_remove_readonly(func, path, exc_info):
    """
    Callback để bỏ chế độ read-only và thử lại.
    """
    # Đặt lại quyền write cho file/folder
    os.chmod(path, stat.S_IWRITE)
    # Gọi lại chính hàm đã lỗi (os.remove hoặc os.rmdir)
    func(path)

def remove_folders_with_no_faces(root_directory):
    for label in os.listdir(root_directory):
        folder_path = os.path.join(root_directory, label)
        if not os.path.isdir(folder_path):
            continue

        jpgs = [f for f in os.listdir(folder_path) if f.lower().endswith('.jpg')]
        if not jpgs:
            continue

        no_face = 0
        for fn in jpgs:
            img = cv2.imread(os.path.join(folder_path, fn))
            if img is None:
                no_face += 1
                continue
            gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
            faces = face_cascade.detectMultiScale(gray,
                                                  scaleFactor=1.1,
                                                  minNeighbors=5,
                                                  minSize=(30, 30),
                                                  flags=cv2.CASCADE_SCALE_IMAGE)
            if len(faces) == 0:
                no_face += 1

        if no_face == len(jpgs):
            # Dùng onerror để tự động bỏ readonly và xóa
            shutil.rmtree(folder_path, onerror=_handle_remove_readonly)

if __name__ == '__main__':
    # Nếu cần chạy với quyền cao hơn, hãy mở cmd/PowerShell as Admin
    remove_folders_with_no_faces(root_dir)
```

```python
import os
import shutil
import random
from pathlib import Path

# --- CẤU HÌNH ---
# Thư mục gốc chứa dữ liệu gốc với các thư mục con là tên từng người
src_root = Path(r"C:\Users\DELL\OneDrive\Documents\FaceDetection\Download_Images\japan2")
# Thư mục đích chứa kết quả Split
dst_root = src_root.parent / "SplitData_by_person"
# Tỉ lệ chia: train, validation, test
split_ratio = {"train": 0.8, "val": 0.1, "test": 0.1}

# Hàm chia dữ liệu theo thư mục người (identity-level split)
def split_by_person(src_root: Path, dst_root: Path, ratio: dict):
    # Lấy danh sách các thư mục con (mỗi thư mục là toàn bộ ảnh của một người)
    persons = [d for d in src_root.iterdir() if d.is_dir()]
    if not persons:
        print(f"Không tìm thấy thư mục con trong {src_root}")
        return

    # Trộn ngẫu nhiên danh sách thư mục người
    random.shuffle(persons)

    # Tính số lượng folder cho từng tập
    total = len(persons)
    n_train = int(total * ratio['train'])
    n_val = int(total * ratio['val'])
    n_test = total - n_train - n_val

    splits = {
        'train': persons[:n_train],
        'val': persons[n_train:n_train + n_val],
        'test': persons[n_train + n_val:]
    }

    # Copy nguyên thư mục mỗi người vào tập tương ứng
    for split_name, person_dirs in splits.items():
        for person_dir in person_dirs:
            target_dir = dst_root / split_name / person_dir.name
            # Copy toàn bộ folder và file con
            shutil.copytree(person_dir, target_dir)
            print(f"Copied {person_dir.name} to {split_name}")

    print(f"Hoàn tất: dữ liệu chia theo thư mục người đã lưu tại {dst_root}")


if __name__ == '__main__':
    # Tạo thư mục đích nếu chưa có
    dst_root.mkdir(parents=True, exist_ok=True)
    split_by_person(src_root, dst_root, split_ratio)
```
