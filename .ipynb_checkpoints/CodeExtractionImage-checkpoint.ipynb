```python
base_dir = "Download_Images"
```

```python
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36",
    "Referer": "https://commons.wikimedia.org/"
}
```

```python
for nationality in df["nationality"].unique():
    save_dir = os.path.join(base_dir, nationality)  
    os.makedirs(save_dir, exist_ok=True)

    df_nation = df[df["nationality"] == nationality]  

    for _, row in df_nation.iterrows():
        image_url = row["image"]
        image_id = row["id"]

        try:
            response = requests.get(image_url, headers=headers, timeout=10)
            if response.status_code == 200:
                file_path = os.path.join(save_dir, f"{image_id}.jpg")
                with open(file_path, "wb") as file:
                    file.write(response.content)
                print(f"Ảnh được tải: {file_path}")
            else:
                print(f"Lỗi tải ảnh {image_id}: {response.status_code}")
        except requests.exceptions.RequestException as e:
            print(f"Không truy cập được link ảnh {image_id}: {e}")
            
print("Completed")
```

```python
df_china = df[df["nationality"] == "china"]

for nationality in df_china["nationality"].unique(): 
    save_dir = os.path.join(base_dir, nationality)  
    os.makedirs(save_dir, exist_ok=True)

    for _, row in df_china.iterrows():  
        image_url = row["image"]
        image_id = row["id"]

        try:
            response = requests.get(image_url, headers=headers, timeout=10)
            if response.status_code == 200:
                file_path = os.path.join(save_dir, f"{image_id}.jpg")
                with open(file_path, "wb") as file:
                    file.write(response.content)
                print(f"Ảnh được tải: {file_path}")
            else:
                print(f"Lỗi tải ảnh {image_id}: {response.status_code}")
        except requests.exceptions.RequestException as e:
            print(f"Không truy cập được link ảnh {image_id}: {e}")

print("Completed")
```
