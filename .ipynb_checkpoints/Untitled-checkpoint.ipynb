```python
from PIL import Image, UnidentifiedImageError
import os

# Đường dẫn thư mục gốc
root_folder = r"C:\Users\DELL\OneDrive\Documents\FaceDetection\Download_Images\japan1"

# Ngưỡng kích thước tối thiểu (width và height)
min_width, min_height = 100, 100

# Duyệt toàn bộ file ảnh trong tất cả thư mục con
for subdir, dirs, files in os.walk(root_folder):
    for filename in files:
        if not filename.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.gif')):
            continue

        img_path = os.path.join(subdir, filename)
        try:
            # Mở ảnh với context manager để tự động đóng file
            with Image.open(img_path) as img:
                # Chuyển sang RGB (loại bỏ alpha/palette nếu có)
                img = img.convert('RGB')
                width, height = img.size

                if width < min_width or height < min_height:
                    # Đóng file trước khi xóa
                    img.close()
                    os.remove(img_path)
                    print(f"Đã xóa ảnh quá nhỏ: {img_path}")
                else:
                    # Resize và lưu giữ định dạng gốc
                    img_resized = img.resize((160, 160), Image.LANCZOS)
                    img_resized.save(img_path, format=img.format)
                    print(f"Đã resize: {img_path}")

        except UnidentifiedImageError:
            print(f"Tập tin không phải ảnh hoặc hỏng: {img_path}")
        except PermissionError:
            print(f"Không có quyền xử lý file: {img_path}")
        except Exception as e:
            print(f"Lỗi khi xử lý {img_path}: {e}")

print("Hoàn tất quá trình lọc, xóa và resize ảnh!")
```

```python
from pathlib import Path
import random
import shutil

# Đường dẫn gốc và đầu ra
root_folder   = Path(r"C:\Users\DELL\OneDrive\Documents\FaceDetection\Download_Images\japan1")
output_folder = Path(r"C:\Users\DELL\OneDrive\Documents\FaceDetection\Download_Images\japan1_split")
extensions     = {'.png', '.jpg', '.jpeg', '.bmp', '.gif'}

# 1. Đệ quy lấy danh sách tất cả file ảnh
all_imgs = [p for p in root_folder.rglob('*') if p.suffix.lower() in extensions]  # :contentReference[oaicite:0]{index=0}

# 2. Trộn ngẫu nhiên và chia tỷ lệ 80%/10%/10%
random.seed(42)
random.shuffle(all_imgs)                                                        # :contentReference[oaicite:1]{index=1}
n_total = len(all_imgs)
n_train = int(0.8 * n_total)
n_val   = int(0.1 * n_total)

splits = {
    'train': all_imgs[:n_train],
    'val':   all_imgs[n_train:n_train + n_val],
    'test':  all_imgs[n_train + n_val:]
}

# 3. Tạo cấu trúc thư mục và copy giữ nguyên hierarchy
for split_name, paths in splits.items():
    for src_path in paths:
        # Lấy đường dẫn tương đối so với gốc
        rel_path = src_path.relative_to(root_folder)                          # :contentReference[oaicite:2]{index=2}
        dst_path = output_folder / split_name / rel_path

        # Tạo thư mục cha nếu chưa tồn tại
        dst_path.parent.mkdir(parents=True, exist_ok=True)
        # Copy file, giữ metadata (timestamp, permissions,…)
        shutil.copy2(src_path, dst_path)                                      # :contentReference[oaicite:3]{index=3}

print(f"Chia xong: train={len(splits['train'])}, val={len(splits['val'])}, test={len(splits['test'])}")
```
