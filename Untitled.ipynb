```python
from PIL import Image, UnidentifiedImageError
import cv2
import os

# Đường dẫn thư mục chứa ảnh
root_folder = r"C:\Users\DELL\OneDrive\Documents\FaceDetection\Download_Images\japan1"

# Đường dẫn tới file haarcascade (bạn cần tải về)
face_cascade_path = r"C:\Users\DELL\OneDrive\Documents\FaceDetection\haarcascade_frontalface_default.xml"

# Kiểm tra file Haar Cascade tồn tại chưa
if not os.path.exists(face_cascade_path):
    raise Exception(f"Không tìm thấy file Haar Cascade tại: {face_cascade_path}")

# Load bộ phát hiện khuôn mặt
face_cascade = cv2.CascadeClassifier(face_cascade_path)

# Ngưỡng kích thước tối thiểu
min_width, min_height = 100, 100

# Xử lý từng file ảnh
for subdir, dirs, files in os.walk(root_folder):
    for filename in files:
        if not filename.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.gif')):
            continue

        img_path = os.path.join(subdir, filename)

        try:
            # Bước 1: Kiểm tra và resize bằng Pillow
            with Image.open(img_path) as img:
                img = img.convert('RGB')
                width, height = img.size

                if width < min_width or height < min_height:
                    print(f"Ảnh quá nhỏ, xóa: {img_path}")
                    img.close()
                    os.remove(img_path)
                    continue

                # Resize ảnh về 160x160
                img_resized = img.resize((160, 160), Image.LANCZOS)
                img_resized.save(img_path, format=img.format)

        except UnidentifiedImageError:
            print(f"Ảnh bị lỗi, xóa: {img_path}")
            try:
                os.remove(img_path)
            except Exception as e:
                print(f"Không thể xóa file lỗi: {e}")
            continue
        except PermissionError:
            print(f"Không có quyền truy cập file: {img_path}")
            continue
        except Exception as e:
            print(f"Lỗi khi xử lý ảnh (PIL): {img_path} - {e}")
            continue

        try:
            # Bước 2: Phát hiện khuôn mặt bằng OpenCV
            img_cv = cv2.imread(img_path)
            if img_cv is None:
                print(f"Ảnh không đọc được (OpenCV), xóa: {img_path}")
                os.remove(img_path)
                continue

            gray = cv2.cvtColor(img_cv, cv2.COLOR_BGR2GRAY)
            faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5)

            if len(faces) == 0:
                print(f"Không phát hiện khuôn mặt, xóa: {img_path}")
                os.remove(img_path)
            else:
                print(f"Phát hiện {len(faces)} khuôn mặt trong: {img_path}")

        except Exception as e:
            print(f"Lỗi khi kiểm tra khuôn mặt: {img_path} - {e}")
            continue

print("✅ Hoàn tất lọc, resize và xóa ảnh lỗi hoặc không có mặt người!")
```
